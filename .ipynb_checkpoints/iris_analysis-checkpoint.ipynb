Task3: We loaded the Iris dataset from data/iris.csv into a Pandas DataFrame named df. We inspected the dataset dimensions, structure, column names, and data types.

```python
rows, cols = df.shape
```

```python
print(f"1.num of rows: {rows} | num of cols: {cols}")
```

```python
print("2.Missing values: ")
```

```python
print(df.isnull().sum())
```

```python
print(f"3.Duplicate rows: {df.duplicated().sum()}")
```

```python
print(f"4. Unique species names: {df['species'].unique()}")
```

```python
print("5.Num of flowers per type:")
```

```python
print(df['species'].value_counts())
```

```python
avg_sepal_length = df['sepal_length'].mean()
```

```python
print(f"6.Average sepal length (sepal_length): {avg_sepal_length:.2f}")
```

```python
max_petal_length = df['petal_length'].max()
```

```python
print(f"7.Maximum petal length (petal_length): {max_petal_length}")
```

```python
min_sepal_width = df['sepal_width'].min()
```

```python
print(f"8.Average sepal length (sepal_width): {min_sepal_width}")
```

```python
print("9.Average characteristics for each type of flower:")
```

```python
display(df.groupby('species').mean())
```

```python
print("10.Descriptive statistics for the data:")
```

```python
print(df.describe())
```

Task4: In this task, we performed an exploratory data analysis on the Iris dataset. We confirmed the dataset size, checked that there are no missing values, and found a few duplicate rows. We also computed overall statistics (mean, min, max) as well as species-specific averages using groupby().
