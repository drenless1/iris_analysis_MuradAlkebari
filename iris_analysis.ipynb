Task4: In this task, we performed an exploratory data analysis on the Iris dataset. We confirmed the dataset size, checked that there are no missing values, and found a few duplicate rows. We also computed overall statistics (mean, min, max) as well as species-specific averages using groupby().

```python
print("1.Display all flowers where the cup length is greater than 6 sepal_length > 6:")
sepal_gt_6 = df[df['sepal_length'] > 6]
display(sepal_gt_6.head())
```

```python
print("2.Display all flowers with a petal width greater than 1.5 petal_width > 1.5:")
petal_gt_1_5 = df[df['petal_width'] > 1.5]
display(petal_gt_1_5.head())
```

```python
print("3.Flowers of the setosa type only:")
setosa_df = df[df['species'] == 'setosa']
display(setosa_df.head())
```

```python
highest_petal_species = df.groupby('species')['petal_length'].mean().idxmax()
highest_petal_value = df.groupby('species')['petal_length'].mean().max()
print(f"4.The species with the highest average petal length is: {highest_petal_species} (mean: {highest_petal_value:.2f})")
```

```python
print("5. The data is sorted by sepal_length in descending order:")
sorted_df = df.sort_values(by='sepal_length', ascending=False)
display(sorted_df.head())
```


Task5: In this task, we performed data filtering and conditional analysis using Pandas:
1. We filtered the dataset for rows where sepal_length > 6 and petal_width > 1.5
2. We extracted flowers belonging strictly to the setosa species
3. We identified that the virginica species has the highest average petal length
4. Finally, we sorted the DataFrame by sepal_length in descending order

```python
import matplotlib.pyplot as plt
species_means = df.groupby('species').mean()
species_means.plot(kind='bar', figsize=(9, 5))
plt.title("Average Iris Features by Species")
plt.xlabel("Species")
plt.ylabel("Measurement (cm)")
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

Following the official Pandas plotting guide, we created a bar chart (kind='bar') to visually compare the average values of each numerical feature across all three Iris species (setosa, versicolor, and virginica)

```python
import os
os.makedirs('output', exist_ok=True)
species_summary = df.groupby('species').mean()
species_summary.to_csv('output/species_summary.csv')
filtered_df = df[df['sepal_length'] > 6]
filtered_df.to_csv('output/filtered_iris.csv', index=False)
```
